# Regla del trapecio

Interpolando dos puntos $(x_{0}, f(x_{0}))$ y $(x_{1}, f(x_{1}))$, mediante un polinomio de Lagrange

\begin{equation*}
    f(x) =
    \begin{bmatrix}
        \cfrac{x - x_{1}}{x_{0} - x_{0}} \\
        \cfrac{x - x_{0}}{x_{1} - x_{0}}
    \end{bmatrix}^{T}
    \begin{bmatrix}
        f(x_{0}) \\
        f(x_{1})
    \end{bmatrix}
\end{equation*}

Integrando

\begin{equation*}
    \int_{x_{0}}^{x_{1}} f(x) \ dx =
    \begin{bmatrix}
        -\cfrac{1}{2} x_{0} + \cfrac{1}{2} x_{1} \\
        -\cfrac{1}{2} x_{0} + \cfrac{1}{2} x_{1}
    \end{bmatrix}^{T}
    \begin{bmatrix}
        f(x_{0}) \\
        f(x_{1})
    \end{bmatrix}
\end{equation*}
